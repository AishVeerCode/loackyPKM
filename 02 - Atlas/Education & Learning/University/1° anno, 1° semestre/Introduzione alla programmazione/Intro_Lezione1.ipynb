### Espressioni e operazioni su numeri
Una espressione è composta da operandi e operatori e, una volta valutata, restituisce un valore.

La più semplice espressione aritmetica è un singolo numero, e.g, 7 (valore restituito 7).

Python distingue tra due tipi di dati numerici:
* numeri interi (int); esempi: 12, -9, 1
* numeri frazionari (float); esempi: 3.14, -45.2, 1.0, -42.3e-4 (notazione esponenziale che rappresenta il numero $−42,3×10^{-4}$)

**Nota bene 1:** Python, come tutti i linguaggi di programmazione, usa lo standard americano, ovvero il punto, per separare la parte intera di un reale da quella frazionaria: 3.14 e non 3,14.

**Nota bene 2:** Al contrario di tutti gli altri linguaggi di programmazione che permettono di rappresentare intervalli di interi limitati (con 64 bit l'intero massimo rappresentabile è 18.446.744.073.709.551.615, dell'ordine di 10^20), in Python è possibile rappresentare interi grandi a piacere. 

Per rappresentare in modo leggibile interi molto grandi tipo 10304100231 in Europa si usa il punto: 10.304.100.231; in America la virgola 10,304,100,231. In python non è possibile usare nessuna delle due opzioni  (10.304.100.231 è un errore, 10,304,100,231 rappresenta 4 interi). Python permette di usare l'underscore nei numeri interi per renderli più leggibili: 10304265181 e 10_304_265_181 sono lo stesso intero ma il secondo è più leggibile e si capisce subito che l'ordine di grandezza è di 10 miliardi. 

Espressione aritmetiche più complesse si ottengono combinando numeri attraverso operatori (addizione, divisione, ecc.), e usando le parentesi tonde per definire la precedenza tra gli operatori. E.g., 3*(5+2) (valore restituito 21).

Gli operatori disponibili nel linguaggio Python sono i seguenti:

Simbolo | Descrizione Operatore 
:---: |:-------: 
\+ 	|  somma
\- 	|  sottrazione
\* 	|  moltiplicazione
/  	|  divisione
// 	|  divisione intera
%  	|  modulo (resto di una divisione)
\**	|  elevamento a potenza


Se entrambi gli operandi di \+, \- e \* sono interi, il risultato è rappresentato come intero, altrimenti è rappresentato come numero frazionario.   

```python
1 + 5
```

```python
1.0 + 5
```

Indipendentemente dal fatto che gli operandi di una divisione / siano interi o frazionari, il quoziente restituito è sempre frazionario

```python
2/5
```

```python
6/2
```

L’operatore // produce sempre *il più grande intero <=  quoziente restituito dalla divisione /*, rappresentato come intero se entrambi gli operandi sono interi, altrimenti come numero frazionario. 

```python
6//3
```

```python
6//4
```

```python
-2//3
```

L’operatore modulo % restituisce il resto della divisione intera. Utile, ad esempio, per vedere se un intero è pari o dispari: n%2 vale 0 se n è pari, 1 se n è dispari.

```python
5%2
```

Ma anche per contare in modo circolare. Ad esempio se vogliamo contare da 0 a 11 in modo circolare (orologio), questo è esattamente il calcolo del resto della divisione per 12

```python
15 % 12
```

% e // sono spesso usati insieme. Per esempio convertire secondi in minuti e secondi rimanenti:




```python
secondi = 137
minuti = secondi // 60   # 2
resto = secondi % 60     # 17
print(secondi,'secondi =',minuti,'minuti e',resto,'secondi')   # 2 min e 17 s
```

### Stringhe
Un altro tipo di dato fondamentale in Python è il tipo *stringa*. Le stringhe sono sequenze ordinate di caratteri delimitati da due *'* o due *"*. I caratteri ammissibili sono i caratteri codificati da UNICODE (praticamente tutti i caratteri di tutti i linguaggi moderni esistenti) più tutti i simboli presenti sulle tastiere o associati a tasti speciali: *@$&/()!⌘⌫*, eccetera. Esempi di stringhe Python corrette:
* 'Ciao'
* "Ciao $$ a tutti"
* "Sono 썐 andato all'interno del negozio ᎈ"
* 'Il mio amico ha detto "Salve a tutti!" con voce sicura'

Attenzione all'uso congiunto di *'* e *"*.
  
UNICODE verrà discusso in una lezione specifica: qui basta sapere che tale standard codifica ciascun carattere o simbolo con un intero che permette di rappresentarlo all'interno del computer.
Per esempio il carattere 'a' è codificato con l'intero 97, il carattere '썐' con l'intero 50000.
La funzione predefinita di python *chr()* permette di scoprire il carattere associato a un intero: *chr(98)* restituisce il carattere 'b', mentre la funzione *ord()* fa l'operazione inversa: *ord('a')* restituisce 97. 

N.B. La codifica di UNICODE definisce un ordinamento (o collazione) tra tutti i caratteri rappresentati permettendo di ordinare anche stringhe composte da lettere maiuscole, minuscole  e caratteri 'strani': per esempio, il computer ordina i file 1_Report.pdf, A_Report.pdf, b_report.pdf, 썐_Report.pdf in quest'ordine perchè '1'<'A'<'b'<'썐', ovvero ord('1') vale 49, ord('A') vale 65, ord('b') vale 98  e ord('썐') vale 50000. 




### Stringhe non corrette
Esempi di stringhe non corrette sono:
1. 'Ciao
2. Ciao a tutti"
3. "Sono andato all'interno del negozio'
4. 'Il mio amico ha detto "Salve a tutti!' con voce sicura"
5. 'Ho aperto l'acqua'

La 1 è sbagliata perché manca l'apice alla fine. Nella 2 manca all'inizio. Nella 3 abbiamo usato apici diversi all'inizio ed alla fine. Nella 4 abbiamo invertito gli apici a destra in modo che Python pensa che la stringa finisca dopo il '!' e non capisce cosa siano i caratteri successivi. Nella 5 abbiamo usato ' all'interno di una stringa delimitata da apici e così, come nel caso 4, Python pensa che la stringa sia terminata dopo la 'l' e non capisce cosa siano i caratteri successivi.

```python
'Il mio amico ha detto "Salve a tutti!' con voce sicura"
```
