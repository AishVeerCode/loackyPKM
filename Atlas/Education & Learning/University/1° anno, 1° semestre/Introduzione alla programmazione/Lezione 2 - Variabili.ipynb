### Variabili
Il concetto di variabile è comune a tutti i linguaggi di programmazione. Una variabile è una zona di memoria in cui viene memorizzato un dato **di un certo tipo**, necessario ai calcoli di un programma. Possiamo riferirci alla zona di memoria con un *nome* a nostra scelta, meglio se significativo, ovvero in grado di suggerire il contenuto della variabile, per esempio `voto` o `media`. Ogni linguaggio di programmazione offre dei **tipi elementari** predefiniti. Per esempio Python prevede, tra gli altri, i tipi elementari `carattere`, `intero`, `reale` e `stringa`.

N.B. Una `stringa`, indicata con una sequenza di caratteri  racchiusa tra due apici, per esempio *'esempio di stringa'*, è una sequenza di caratteri e la prossima lezione descriverà il tipo stringa in dettaglio.

La dichiarazione delle variabili *varia* da linguaggio a linguaggio. Per esempio, il C prevede la dichiarazione esplicita di *tutte* le variabili che saranno usate nel programma associando a ogni variabile un tipo che *non* può variare durante l'esecuzione. Per esempio:

**C**
```C
int i;  // i è di tipo intero e il suo tipo NON può cambiare
i = 5;
i = 3.9; // 3.9 viene convertito in intero per troncamento, i vale 3
```
In python, invece, le variabili possono essere introdotte al momento del loro uso e il tipo può cambiare (con possibile perdita di leggibilità del programma):

**Python**
```python
i=5   #creo la variabile i che è di tipo intero (perchè 5 è un intero)
i=3.9 #i ora è di tipo reale e vale 3.9 #operazione sconsigliata
#oppure
i: int #i è di tipo intero e il suo tipo può cambiare...
i=5
i=3.9 
```
In python è possibile inserire un valore in una variabile (con `=` o `input()`), mostrarne il contenuto a schermo con `print()` o utilizzarne il contenuto per fare dei calcoli. Per esempio, volendo calcolare e mostrare l'area di un cerchio possiamo usare due variabili che chiamiamo `raggio` e `area`:

````python
raggio = 2.0 
area = raggio * raggio * 3.14
print(area) # stampa a schermo il risultato
````


### Istruzione di assegnazione `=`
Le istruzioni come `area = raggio*raggio*3.14` si chiamano *istruzioni di assegnazione*. A sinistra del segno `=` abbiamo il nome di una *variabile* (`area` in questo caso) e a destra una *espressione*. La variabile non è altro che un modo per indicare l'indirizzo di una zona di memoria. L'espressione a destra viene valutata (calcolato il suo valore) e il risultato viene assegnato alla variabile a sinistra (viene scritto nella zona di memoria il cui indirizzo è associato al nome della variabile).
Se nell'espressione di destra compare una variabile, viene letto dalla memoria il suo valore attuale.



In una assegnazione si distinguono: 
- il valore `sinistro` (a sinistra di =) che corrisponde a un *indirizzo* di memoria e 
- il valore `destro` (a destra di =) che corrisponde al *valore* dell'espressione che verrà memorizzato nell'indirizzo di memoria. 

Nell'esempio:

area = raggio\*raggio\*3.14 

il valore sinistro `area`  vale 9, ovvero l'indirizzo in cui scrivere il valore dell'espressione; il valore destro `raggio*raggio*3.14` vale 12.56 e viene memorizzato nell'indirizzo corrispondente ad area.



#### L'istruzione di assegnazione NON va confusa con il simbolo matematico di uguaglianza, anche se il simbolo usato è lo stesso `=`

In matematica l'uguaglianza è una relazione binaria di equivalenza che specifica che il valore dell'espressione a sinistra dell'= è uguale a quello dell'espressione a destra. Usando lo stesso simbolo in Python stiamo invece *assegnando* il valore dell'espressione di destra alla variabile che compare a sinistra. Per apprezzare la differenza, notate (per esempio) che l'uguaglianza matematica è simmetrica (cioè dire `a=b` o `b=a` è la stessa cosa) mentre la stessa proprietà non vale per l'assegnazione dove la parte sinistra e la destra hanno un ruolo molto diverso (`area = raggio\*raggio*3.14` è corretta, mentre `raggio\*raggio*3.14 = area` non ha alcun senso). Vedremo in seguito che Python, come quasi tutti i linguaggi di programmazione, usa un simbolo diverso, `==`,  per effettuare un confronto di uguaglianza fra due espressioni.

Nell'esempio python appena mostrato `raggio` e `area` sono due  variabile di tipo `reale` perchè il valore destro dell'assegnazione è un reale. Ma python permette di cambiare il tipo di una variabile durante l'esecuzione di un programma
```python
i=5   #i è di tipo intero
i=3.9 #i ora è di tipo reale e vale 3.9
```
Il linguaggio C, invece richiede di dichiarare il tipo di una variabile prima di usarla:
```C
int i ;  // i è di tipo intero e il suo tipo NON può cambiare
i = 5 ;
i = 3.9; // 3.9 viene convertito in intero per troncamento, i vale 3
```

### Regole per i nomi delle variabili
I nomi delle variabili in Python devono rispettare queste regole:
1. Il primo carattere deve essere una lettera o il simbolo \_ (underscore o trattino in basso)
2. I caratteri successivi possono essere lettere, cifre o il carattere \_
3. Python **distingue** tra maiuscole e minuscole (ad esempio, `area` e `Area` sono due variabili distinte)

### input (), int() e float ()
Nell'esempio di prima abbiamo memorizzato il valore '2.0' nella variabile raggio e il programma  
```python
raggio = 2.0
area = raggio * raggio * 3.14
print(area) 
```
calcola sempre e solo l'area di un cerchio di raggio 2.0.

Per avere un programma 'parametrico', ovvero in cui il raggio sia un parametro che può essere modificato, usiamo l'istruzione input() che permette, ad ogni esecuzione del programma, di leggere da tastiera una stringa e assegnarla a una variabile (dopo averla convertita nel tipo desiderato, reale nell'esempio):

````python
raggio = input('Inserisci il valore del raggio: ')  #raggio è una stringa
raggio = float(raggio)                             #raggio è un reale
area = raggio * raggio * 3.14
print('raggio =',raggio,'area =',area) # stampa a schermo il risultato
````

### Modifica del valore di una variabile
L'assegnazione di un valore a una variabile si può fare anche se la variabile ha già un valore. In questo caso il vecchio valore viene cancellato (perso) e sostituito da quello nuovo (in realtà se cambiate tipo Python cerca una nuova zona di memoria e la associa allo stesso nome)

````python
area=5
print("ora area vale:",area)
area=99
print("dopo l'assegnazione area vale :",area)
````

#### Operatori di assegnazione composti

E’ anche possibile scrivere enunciati che effettuano una assegnazione e svolgono allo stesso tempo un calcolo.

`i /= 3` | `i = i / 3`

### Eseguire più volte un programma
Nella grande maggioranza dei casi, un programma non viene scritto per essere eseguito una sola volta, ma più volte e con dati di ingresso diversi. Nell'esempio precedente, se vogliamo ri-eseguire il programma su valori diversi (per esempio assegnando alla variabile `var1` il valore 'prova' invece di 'ciao') dovremmo modificare il programma prima di eseguirlo.

### Funzione `input()`
Una soluzione migliore è di scrivere un programma che NON assegna un valore fisso alla variabile `var1` ma che invece chiede all'utente di fornirlo al momento dell'esecuzione. In questo modo il programma rimane lo stesso, ma ogni esecuzione avrà diversi dati di ingresso. Il modo più semplice di ottenere questo risultato in Python è usando la funzione `input()`

All'interno dell'Editor sostituiamo la riga
`var1 = 'ciao'`
con
`var1 = input()`
ottenendo il programma:

`var1 = input()`

`var2 = var1*4`

`print(var2)`

ed eseguiamo `Run`.

Il programma non genera alcun risultato perché è in attesa che venga inserito il dato di ingresso *nella Shell*  che verrà poi assegnato alla variabile `var1`. Spostate quindi il mouse nella Python Shell ed inserite il dato, ad esempio: palla

Sulla Shell apparirà il risultato, cioè `pallapallapallapalla` nel nostro caso.

ottenendo il programma:

### La funzione `int()`
Quando inserite un dato da tastiera, Python legge *esattamente* i caratteri che battete, senza preoccuparsi se siano caratteri alfabetici o numerici. Dal punto di vista di qualunque linguaggio di programmazione, incluso Python, *tutto* quello che inserite da tastiera viene inserito in una stringa. Se infatti eseguite il programma precedente e inserite come dato di ingresso 5, il programma stamperà sulla Shell 5555, cioè 4 volte il *carattere* 5, non 20.

Se quello che volete è che  `var1` contenga un numero intero, allora dovete aggiungere la riga

`var1 = int(var1)`

dopo la riga

`var1 = input('Inserisci un dato di ingresso: ')` 

ottenendo il programma:

`var1 = input('Inserisci un dato di ingresso: ')`

`var1 = int(var1)`

`var2 = var1*4`

`print(var2)`

L'effetto della nuova riga è di trasformare il valore della variabile `var1` in un numero intero. Questo è possibile solo se *tutti i caratteri* di `var1` dopo l'istruzione `input()` sono cifre, altrimenti il programma andrà in errore. Provate ad eseguire il nuovo programma **sia nella cella sottostante che nell'Editor** inserendo come dato in ingresso una volta un numero ed una volta una stringa di caratteri alfabetici come ad esempio 'palla'

````python
n = int(input('Inserisci un intero: '))
````

### La funzione `float()`

Ha un funzionamento analogo alla funzione `int()`, ma converte la stringa in input in un numero frazionario. Se la stringa in input non è convertibile in un numero frazionario, viene restituito un errore. 


````python
r = float(input('Inserisci un reale: '))
````

````python
4 + float('3.55')
````

````python
float('3,55')
````

````python
float("12*3")
````

### Riassumendo

Qualunque cosa venga scritta sulla tastiera dopo un comando input() viene **sempre** interpretata come una stringa e occorrono le funzioni int() e float() per trasformarla nell'intero corrispondente. Per esempio,  55 scritto da tastiera corrisponde alla stringa '55' e per convertirla in un intero occorre usare int(). 

*Al contrario*, all'interno di una espressione 55 scritto su una linea di codice corrisponde all'intero 55 e per convertirlo in stringa occorre scrivere "55" o str(55): 

````python
var=55
print(var*3)
var='55'
print(var*3)
````
