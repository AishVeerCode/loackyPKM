## Istruzione For
Tutti i linguaggi di programmazione prevedono un'istruzione `for`, ovvero una istruzione di ciclo, per ripetere un blocco di istruzioni un certo numero prefissato di volte. La sintassi di tale istruzione e il modo di definire sia il numero di ripetizioni che il blocco di istruzioni variano, però, in base al linguaggio utilizzato. 

Python prevede l'uso esplicito di una `sequenza di valori` che viene poi visitata valore per valore, eseguendo un numero di cicli pari alla `lunghezza della sequenza`.

L'istruzione `for` ha la seguente struttura in Python

```python
for i in sequenza_di_valori:
    istruzione1
    istruzione2
    ..
    istruzionen
istruzione successiva al for
```

dove `i` è una variabile, detta variabile di ciclo (il nome della variabile non è fisso, lo scegliamo noi) e la sequenza di valori può essere definita usando una variabile del programma, una costante definita nel for, o usando la funzione `range`.

L'esecuzione avviene nel seguente modo: La variabile `i` assume il primo valore della `sequenza_di_valori` e vengono eseguite tutte le istruzioni da 1 a n. Poi la variabile `i` assume il secondo valore della `sequenza_di_valori` e vengono eseguite tutte le istruzioni da 1 a n, e così via fino a che la `i` assume l'ultimo valore della `sequenza_di_valori`. A questo punto vengono eseguite un'ultima volta tutte le istruzioni da 1 a n e il ciclo `for` termina.

Dal punto di vista sintattico, si noti l'indentazione dopo i due punti e la sua eliminazione al termine dell'istruzione `for`. In Python, un "blocco" di istruzioni si identifica allineando verticalmente tutte le istruzioni che lo compongono, che sono quindi precedute tutte dallo stesso numero di caratteri di tabulazione. Pertanto, per specificare l'istruzione da eseguire dopo che si è usciti dal ciclo `for`, è sufficiente eliminare una indentazione, ovvero scriverla in modo tale che sia allineata verticalmente alla parola chiave `for`. 

### Primi esempi di ciclo for
Le stringhe sono sequenze di caratteri e quindi è possibile utilizzare una stringa per ripetere un blocco di istruzioni un numero di volte pari alla lunghezza della stringa usando un ciclo `for`

````python
s = 'casetta' #una stringa è una sequenza di caratteri

for c in s:  #per ogni carattere (c) in s
    print(c) #stampa il carattere e vai a riga nuova
````

La sequenza può essere indicata direttamente come costante nel for:

````python
for c in 'pippo':
    print(c) #stampa il carattere e vai a riga nuova
````

Non è affatto obbligatorio usare la variabile di ciclo nel for:

````python
for c in 'pippo':
    print(6) 
````

La variabile di ciclo può essere usata per fare calcoli:

````python
for c in 'pippo':
    print(c,ord(c)) #stampa il valore UNICODE dei caratteri della stringa
````

### Python vs  C 
Nel linguaggio C (oggetto di un prossimo corso) la sintassi del for, la logica che determina il numero di iterazioni, e il modo di indicare un blocco di istruzioni sono completamente differenti: il seguente codice python che stampa i numeri da 1 a 5:

```python
for i in [1,2,3,4,5]:
    print(i)
```
in C viene scritto così:

```C
int i;
for (i = 1; i <= 5; i=i+1) {
    printf("%d\n", i);
}
```
Trascurando la sintassi della **printf**, che non è opportuno discutere qui; basta sapere che stampa il valore di *i* e va a capo, si vogliono evidenziare le seguenti differenze:

1. in C le variabili vanno dichiarate prima di poterle usare indicandone il tipo: i è un variabile di tipo intero, tipo che **non** può essere cambiato all'interno del programma

2. ogni istruzione in C deve essere chiusa con un **;**

3. il blocco di istruzioni è delimitato da due parentesi graffe e l'indentazione serve solo a rendere il codice più leggibile. Il seguente codice:

```C
int i; for (i = 1; i <= 5; i=i+1) {printf("%d\n", i);}
```
viene eseguito perfettamente mentre
```python
for i in [1,2,3,4,5]: print(i)
```
genera un errore

4. nel C non esiste una **sequenza** e la logica delle tre istruzioni tra parentesi *(i = 1; i <= 5; i=i+1)* dopo il for è:
    1. inizializzazione i = 1, prima di entrare nel for i vale 1
    2. i <=5 condizione che deve essere vera prima di eseguire una iterazione 
    3. istruzione sempre eseguita alla fine di ogni iterazione i=i+1
  
Il for del C è di fatto una forma compatta di **while** (che vedremo nelle prossime lezioni), e il programmatore gestisce ogni aspetto del ciclo

### Sequenze numeriche create con la funzione `range()`
E se volessimo stampare i numeri tra 1 e 100? In C basta modificare i<=5 dell'esempio sopra in i<=100. E in python? Posso creare una lista [1,2,3,..., 100] ma è ovvio che non è la soluzione ottimale. Python usa la funzione `range()` per definire una sequenza numerica da poter poi usare nel `for`.  Se usiamo la funzione con un solo parametro  `n` (`range(n)`) essa genera la sequenza di tutti i numeri naturali da `0` (incluso) ad `n` (escluso).

````python
range(10)
````

Per visualizzare effettivamente la sequenza di valori generata dalla funzione `range()` la convertiremo in una lista con la funzione `list()`

````python
list(range(10))
````

Per generare sequenze che non partono da 0, possiamo usare `range(n1,n2)` che invece genera la sequenza di tutti i numeri naturali da `n1` (incluso) ad `n2` (escluso).

````python
list(range(1,11))
````

Quindi il ciclo for per stampare tutti i numeri tra 1 e 100 è:

```python
for i in range(1,101):
    print(i)
```
Un'ultima variante prevede di specificare 3 parametri, in cui il primo è il valore iniziale (incluso), il secondo il valore finale (escluso) ed il terzo è il passo (step) usato per andare dal valore iniziale a quello finale. Notate che il passo può anche essere negativo, in questo caso il valore iniziale deve essere maggiore di quello finale.

````python
list(range(5,22,3))
````

````python
list(range(21,5,-2))
````

Ricorda che quando abbiamo visto le stringhe, abbiamo introdotto l'operazione di slicing, la cui sintassi, nella versione più generale è `s[i:j:n]`. Tale operazione è concettualmente simile alla funzione `range`, ovvero restituisce tutti e soli i caratteri di `s` nelle posizioni specificate nella sequenza generata da `range(i,j,n)`.

````python
s="paperopoli"
print(s,'s[1:8:2]',s[1:8:2],'indici', list(range(1,8,2)))
print(s,'s[0:10:3]',s[0:10:3],'indici', list(range(0,10,3)))
print(s,'s[8:2:-2]',s[8:2:-2],'indici', list(range(8,2,-2)))
print(s,'s[::-1]',s[::-1],'indici', list(range(len(s)-1,-1,-1))) 
````

### Ciclo sui caratteri di una stringa (indice o carattere)
Come abbiamo già visto, è possibile effettuare un ciclo `for` che scandisce tutti i caratteri di una stringa in 2 modi diversi, con un ciclo sugli indici oppure con un ciclo sui caratteri. Vediamo differenze, vantaggi e svantaggi delle 2 soluzioni.

Vantaggi del ciclo sui caratteri:
1. Codice più semplice ed immediato
2. Accesso diretto ai caratteri

Svantaggi:
1. Non conoscenza della posizione dei caratteri
2. Difficoltà a scandire in ordine diverso da quello standard

````python
#scansione stringa tramite accesso diretto ai suoi elementi
#ovvero la stringa è una "sequenza" di caratteri
s=input("Inserisci una stringa: ")

for c in s:
    print(c,"=",ord(c))

print()
    
for i in range(len(s)):
    print(s[i],"=",ord(s[i]))
````

#### Esempio: stampa dei caratteri di una stringa in ordine inverso
Scrivere un programma che legge una stringa e stampa i suoi caratteri in ordine inverso

````python
#stampa dei caratteri di una stringa in ordine inverso
#tramite indice. 
s=input('Inserisci la stringa: ')

for i in range(len(s)-1,-1,-1):
#genera l'insieme dei valori da len(s) - 1 fino a 0 (-1 escluso)
#in ordine decrescente (con passo -1)
    print(s[i])

print()
    
# se scandiamo i caratteri di una stringa, non possiamo in effetti 
# cambiare l'ordine di scansione, come invece fatto usando gli indici. 
# In questo esempio, comunque potremmo prima invertire la stringa 
# per poi scandirla successivamente. Ciò non toglie che la scansione 
# tramite indici ci offre maggiore flessibilità

for c in s[::-1]:
    print(c)
````

### Variabili di tipo accumulatore


Un uso molto frequente delle variabili è di usarle per memorizzare valori che cambiano durante l'esecuzione del programma. In molti problemi, non siamo in grado di calcolare direttamente la soluzione, ma la otteniamo attraverso vari passaggi duranti i quali la variabile memorizza un valore parziale, che solo alla fine diventa quello definitivo.

### Cicli con accumulatori

Spesso in un ciclo si ha la necessità di *accumulare* i valori di una sequenza o insieme usando una variabile, chiamata appunto accumulatore. Il significato di accumulare dipende dall'obiettivo della computazione che il ciclo deve effettuare. Ad esempio, può significare effettuare una somma, o una moltiplicazione, o una concatenazione di stringhe, ecc. Alla fine del ciclo la variabile usata come accumulatore contiene il valore finale che si intendeva calcolare attraverso il ciclo. In genere si accumula *fino alla scansione completa di una sequenza*, o *fino al raggiungimento di un certo valore* per l'accumulatore (vedi esempi successivi).

L'accumulatore deve essere inizializzato prima di entrare nel ciclo. In genere è inizializzato ad un valore neutro rispetto all'operazione utilizzata per accumulare dati. Se ad esempio vogliamo accumulare numeri (interi o frazionari) eseguendo delle somme, l'accumulatore dovrà essere inizializzato a `0`. Similmente, se l'operazione di accumulazione fosse una moltiplicazione di numeri, si dovrebbe inizializzare l'accumulatore al valore `1`.

#### Esempio: somma dei numeri interi compresi tra 2 estremi
Scrivere un programma che prende in input 2 numeri interi n1 e n2 letti da tastiera e calcola la somma di tutti i numeri compresi tra n1 e n2 (estremi inclusi).

````python
n1 = int(input("Inserisci l'estremo inferiore: "))
n2 = int(input("Inserisci l'estremo superiore: "))
somma=0

for num in range(n1,n2+1):
    somma=somma+num

print("somma = ",somma)
````

#### Esempio: statistiche su generatore random
Scrivere un programma che prende in input un numero intero n ed il numero rip di ripetizioni da effettuare e genera rip numeri casuali tra 0 ed n e ne calcola il valore medio

````python
from random import randint
somma=0
n = int(input("Inserisci il limite di variabilità: "))
rip = int(input("Inserisci il numero di ripetizioni: "))

for i in range(rip):
    num = randint(0,n)
    somma += num

print("valore medio dei numeri casuali = ",somma/rip)
````

#### Esempio: accumulatore con le stringhe: sostituzione caratteri di una stringa con il successivo nella tabella UNICODE
Il seguente programma legge da tastiera una stringa `s` e stampa a schermo una nuova stringa in cui ogni carattere di `s` è sostituito con il carattere successivo nella tabella UNICODE.

````python
s = input('immetti una stringa: ')

s1=''

for c in s:
    prox = chr(ord(c)+1)
    s1 += prox
    
print(s1)
````
